```python
import requests

# WeatherAPI Key (Replace with your actual API key)
API_KEY = "7ef0cc2edc5e4ed8893101148252402"
CITY = "London"
BASE_URL = f"https://api.weatherapi.com/v1/current.json?key={API_KEY}&q={CITY}"

# API Request
response = requests.get(BASE_URL)

# Check if request was successful
if response.status_code == 200:
    data = response.json()
    weather = data["current"]["condition"]["text"]
    temperature = data["current"]["temp_c"]
    humidity = data["current"]["humidity"]
    wind_speed = data["current"]["wind_kph"]

    print(f"Weather in {CITY}: {weather}")
    print(f"Temperature: {temperature}°C")
    print(f"Humidity: {humidity}%")
    print(f"Wind Speed: {wind_speed} km/h")
else:
    print(f"Error {response.status_code}: Unable to fetch weather data")
```
